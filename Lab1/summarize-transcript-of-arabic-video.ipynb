```python
from urllib.parse import urlparse, parse_qs
from youtube_transcript_api import YouTubeTranscriptApi

def extract_video_id(url: str) -> str:
     """
       Extract the YouTube video ID from a URL.
       Raises ValueError if no 'v' parameter is found.
     """
     parsed = urlparse(url)
     qs = parse_qs(parsed.query)
     video_ids = qs.get('v')

     if not video_ids:
        raise ValueError(f"No video id found in URL: {url}")
     return video_ids[0]

if __name__ == "__main__":
          url = "https://www.youtube.com/watch?v=-INbpJksLro"

          video_id = extract_video_id(url)

          api = YouTubeTranscriptApi()
          fetched = api.fetch(video_id, languages=['ar']) # returns a FetchedTranscript

          text = "\n".join(snippet.text for snippet in fetched)
```

```python
def chunk_text(text, max_length=1000, overlap=200):
    chunks = []
    start = 0
    
    while start < len(text):
        end = start + max_length
        chunk = text[start:end]
        chunks.append(chunk)
        start = end - overlap
    
    return chunks
```

```python
from transformers import BertTokenizer, AutoModelForSeq2SeqLM, pipeline
from arabert.preprocess import ArabertPreprocessor

model_name="malmarjeh/mbert2mbert-arabic-text-summarization"
preprocessor = ArabertPreprocessor(model_name="")

tokenizer = BertTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
pipeline = pipeline("text2text-generation",model=model,tokenizer=tokenizer)
```

```python
# preprocess full text
processed_text = preprocessor.preprocess(text)

# split into chunks
chunks = chunk_text(processed_text)
print("Number of chunks:", len(chunks))
```

```python
chunk_summaries = []

for i, chunk in enumerate(chunks):
    summary = pipeline(
        chunk,
        pad_token_id=tokenizer.eos_token_id,
        num_beams=3,
        repetition_penalty=3.0,
        max_length=120,
        length_penalty=1.0,
        no_repeat_ngram_size=3
    )[0]['generated_text']
    
    chunk_summaries.append(summary)
    print(f"Chunk {i+1} summarized")
```

```python
combined_summary = " ".join(chunk_summaries)
```

```python
final_summary = pipeline(
    combined_summary,
    pad_token_id=tokenizer.eos_token_id,
    num_beams=4,
    repetition_penalty=2.5,
    max_length=200,
    length_penalty=1.2,
    no_repeat_ngram_size=3
)[0]['generated_text']

print("Final Summary:\n", final_summary)
```
